**Investigating the Bar Coupons**

Now, we will lead you through an exploration of just the bar related coupons.  

1. Create a new `DataFrame` that contains just the bar coupons.


```python
# Create a DataFrame containing only bar coupons
bar_coupons = data[data['coupon'] == 'Bar']
print('Bar coupons shape:', bar_coupons.shape)
```

2. What proportion of bar coupons were accepted?


```python
# Proportion of bar coupons that were accepted
bar_acceptance = bar_coupons['Y'].mean()
print(f'Proportion of bar coupons accepted: {bar_acceptance:.4f} ({bar_acceptance*100:.2f}%)')
```

3. Compare the acceptance rate between those who went to a bar 3 or fewer times a month to those who went more.


```python
# Compare acceptance rate: went to bar 3 or fewer times vs more than 3 times a month
three_or_fewer = bar_coupons[bar_coupons['Bar'].isin(['never', 'less1', '1~3'])]
more_than_three = bar_coupons[bar_coupons['Bar'].isin(['4~8', 'gt8'])]

rate_3_or_fewer = three_or_fewer['Y'].mean()
rate_more_than_3 = more_than_three['Y'].mean()

print(f'Acceptance rate (3 or fewer visits/month): {rate_3_or_fewer:.4f} ({rate_3_or_fewer*100:.2f}%)')
print(f'Acceptance rate (more than 3 visits/month): {rate_more_than_3:.4f} ({rate_more_than_3*100:.2f}%)')
```

4. Compare the acceptance rate between drivers who go to a bar more than once a month and are over the age of 25 to the all others.  Is there a difference?


```python
# Compare acceptance rate: go to bar more than once a month AND over age 25 vs all others
# "more than once a month" = 1~3, 4~8, gt8
# "over 25" = age in ['26', '31', '36', '41', '46', '50plus']
bar_more_than_once = bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])
over_25 = bar_coupons['age'].isin(['26', '31', '36', '41', '46', '50plus'])

group1 = bar_coupons[bar_more_than_once & over_25]
group2 = bar_coupons[~(bar_more_than_once & over_25)]

rate_group1 = group1['Y'].mean()
rate_group2 = group2['Y'].mean()

print(f'Acceptance rate (bar >1/month AND age >25): {rate_group1:.4f} ({rate_group1*100:.2f}%)')
print(f'Acceptance rate (all others):               {rate_group2:.4f} ({rate_group2*100:.2f}%)')
print(f'\nDifference: {abs(rate_group1 - rate_group2)*100:.2f} percentage points')
```

5. Use the same process to compare the acceptance rate between drivers who go to bars more than once a month and had passengers that were not a kid and had occupations other than farming, fishing, or forestry.


```python
# Compare acceptance rate: go to bars more than once a month AND passengers not a kid
# AND occupation not farming, fishing, or forestry
bar_more_once = bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])
not_kid = bar_coupons['passanger'] != 'Kid(s)'
not_farming = bar_coupons['occupation'] != 'Farming Fishing & Forestry'

group_target = bar_coupons[bar_more_once & not_kid & not_farming]
group_rest = bar_coupons[~(bar_more_once & not_kid & not_farming)]

rate_target = group_target['Y'].mean()
rate_rest = group_rest['Y'].mean()

print(f'Acceptance rate (bar >1/month, no kids, not farming/fishing/forestry): {rate_target:.4f} ({rate_target*100:.2f}%)')
print(f'Acceptance rate (all others): {rate_rest:.4f} ({rate_rest*100:.2f}%)')
```

6. Compare the acceptance rates between those drivers who:

- go to bars more than once a month, had passengers that were not a kid, and were not widowed *OR*
- go to bars more than once a month and are under the age of 30 *OR*
- go to cheap restaurants more than 4 times a month and income is less than 50K.



```python
# Compare acceptance rates for the combined group:
# - go to bars more than once a month, passengers not a kid, and not widowed OR
# - go to bars more than once a month and under age 30 OR
# - go to cheap restaurants more than 4 times a month and income < 50K

bar_more_once = bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])
not_kid = bar_coupons['passanger'] != 'Kid(s)'
not_widowed = bar_coupons['maritalStatus'] != 'Widowed'
under_30 = bar_coupons['age'].isin(['below21', '21', '26'])
cheap_restaurant_gt4 = bar_coupons['RestaurantLessThan20'].isin(['4~8', 'gt8'])
low_income = bar_coupons['income'].isin([
    'Less than $12500', '$12500 - $24999', '$25000 - $37499', '$37500 - $49999'
])

cond1 = bar_more_once & not_kid & not_widowed
cond2 = bar_more_once & under_30
cond3 = cheap_restaurant_gt4 & low_income

combined = bar_coupons[cond1 | cond2 | cond3]
others = bar_coupons[~(cond1 | cond2 | cond3)]

rate_combined = combined['Y'].mean()
rate_others = others['Y'].mean()

print(f'Acceptance rate (combined group): {rate_combined:.4f} ({rate_combined*100:.2f}%)')
print(f'Acceptance rate (all others):     {rate_others:.4f} ({rate_others*100:.2f}%)')
```

7.  Based on these observations, what do you hypothesize about drivers who accepted the bar coupons?

**Hypothesis about drivers who accepted bar coupons:**

Based on the analysis above, drivers who accepted bar coupons tend to share several characteristics:

1. **Frequency matters most**: Drivers who visit bars more than once a month have significantly higher acceptance rates than those who rarely or never go. Regular bar-goers are already inclined toward bar visits, making them more receptive to coupons.

2. **Age plays a role**: Younger drivers (under 30) and those over 25 who are regular bar visitors both show higher acceptance. This suggests that younger adults and established bar patrons are the key demographics.

3. **Passenger context**: Drivers without kid passengers are more likely to accept bar coupons, which makes practical sense — a bar is not a family-friendly destination.

4. **Income and dining habits**: Even drivers with lower incomes who frequently eat at cheap restaurants show higher acceptance, indicating that price-conscious but socially active individuals are open to bar coupons.

Overall, the typical bar coupon acceptor is a socially active individual who regularly visits bars, is not driving with children, and tends to be younger or a frequent diner.

### Independent Investigation: Coffee House Coupons

Coffee House coupons are the most common coupon type in the dataset. Let's explore what characteristics distinguish drivers who accept coffee house coupons from those who don't.

```python
# Filter for Coffee House coupons
coffee = data[data['coupon'] == 'Coffee House']
coffee_acceptance = coffee['Y'].mean()
print(f'Coffee House coupon acceptance rate: {coffee_acceptance:.4f} ({coffee_acceptance*100:.2f}%)')
print(f'Total coffee house coupon observations: {len(coffee)}')

# Acceptance rate by coffee house visit frequency
print('\nAcceptance rate by Coffee House visit frequency:')
for freq in ['never', 'less1', '1~3', '4~8', 'gt8']:
    subset = coffee[coffee['CoffeeHouse'] == freq]
    if len(subset) > 0:
        rate = subset['Y'].mean()
        print(f'  {freq:>8s}: {rate:.4f} ({rate*100:.2f}%) - n={len(subset)}')
```

```python
# Visualize acceptance rate by key factors
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Acceptance by coffee house visit frequency
freq_order = ['never', 'less1', '1~3', '4~8', 'gt8']
freq_rates = coffee.groupby('CoffeeHouse')['Y'].mean().reindex(freq_order)
axes[0, 0].bar(freq_rates.index, freq_rates.values, color='steelblue', edgecolor='black')
axes[0, 0].set_title('Acceptance Rate by Coffee House Visit Frequency')
axes[0, 0].set_xlabel('Visit Frequency')
axes[0, 0].set_ylabel('Acceptance Rate')

# 2. Acceptance by time of day
time_order = ['7AM', '10AM', '2PM', '6PM', '10PM']
time_rates = coffee.groupby('time')['Y'].mean().reindex(time_order)
axes[0, 1].bar(time_rates.index, time_rates.values, color='coral', edgecolor='black')
axes[0, 1].set_title('Acceptance Rate by Time of Day')
axes[0, 1].set_xlabel('Time')
axes[0, 1].set_ylabel('Acceptance Rate')

# 3. Acceptance by passenger type
pass_rates = coffee.groupby('passanger')['Y'].mean().sort_values(ascending=False)
axes[1, 0].bar(pass_rates.index, pass_rates.values, color='mediumseagreen', edgecolor='black')
axes[1, 0].set_title('Acceptance Rate by Passenger Type')
axes[1, 0].set_xlabel('Passenger')
axes[1, 0].set_ylabel('Acceptance Rate')

# 4. Acceptance by age group
age_order = ['below21', '21', '26', '31', '36', '41', '46', '50plus']
age_rates = coffee.groupby('age')['Y'].mean().reindex(age_order)
axes[1, 1].bar(age_rates.index, age_rates.values, color='mediumpurple', edgecolor='black')
axes[1, 1].set_title('Acceptance Rate by Age Group')
axes[1, 1].set_xlabel('Age')
axes[1, 1].set_ylabel('Acceptance Rate')

plt.tight_layout()
plt.show()
```

```python
# Deeper comparisons
# 1. Frequent coffee house visitors (>1/month) vs infrequent
frequent = coffee[coffee['CoffeeHouse'].isin(['1~3', '4~8', 'gt8'])]
infrequent = coffee[coffee['CoffeeHouse'].isin(['never', 'less1'])]
print(f'Frequent visitors (>1/month) acceptance rate: {frequent["Y"].mean():.4f} ({frequent["Y"].mean()*100:.2f}%)')
print(f'Infrequent visitors acceptance rate:          {infrequent["Y"].mean():.4f} ({infrequent["Y"].mean()*100:.2f}%)')

# 2. Acceptance by weather and temperature
print('\nAcceptance rate by weather:')
for w in coffee['weather'].unique():
    subset = coffee[coffee['weather'] == w]
    print(f'  {w:>8s}: {subset["Y"].mean():.4f} ({subset["Y"].mean()*100:.2f}%)')

print('\nAcceptance rate by temperature:')
for t in sorted(coffee['temperature'].unique()):
    subset = coffee[coffee['temperature'] == t]
    print(f'  {t}F: {subset["Y"].mean():.4f} ({subset["Y"].mean()*100:.2f}%)')

# 3. Acceptance by expiration
print('\nAcceptance rate by coupon expiration:')
for exp in coffee['expiration'].unique():
    subset = coffee[coffee['expiration'] == exp]
    print(f'  {exp}: {subset["Y"].mean():.4f} ({subset["Y"].mean()*100:.2f}%)')

# 4. Acceptance by destination
print('\nAcceptance rate by destination:')
for dest in coffee['destination'].unique():
    subset = coffee[coffee['destination'] == dest]
    print(f'  {dest}: {subset["Y"].mean():.4f} ({subset["Y"].mean()*100:.2f}%)')
```

### Findings: Coffee House Coupons

**Key observations:**

1. **Visit frequency is the strongest predictor**: Drivers who already visit coffee houses regularly (more than once a month) accept coupons at a substantially higher rate than those who rarely visit. This mirrors the pattern we saw with bar coupons.

2. **Time of day matters**: Morning coupons (10AM) tend to have higher acceptance rates than evening coupons, aligning with typical coffee consumption habits.

3. **Passenger type influences acceptance**: Drivers with friends tend to accept coffee house coupons more readily, likely because a coffee house is a social destination. Those driving alone or with kids show lower acceptance.

4. **Younger drivers are more receptive**: Drivers under 30 show higher acceptance rates, suggesting that younger demographics are more spontaneous or frequent coffee consumers.

5. **Coupon expiration**: Coupons with a 1-day expiration window have higher acceptance rates than those expiring in 2 hours, giving drivers more flexibility to plan a visit.

6. **Destination**: Drivers with no urgent destination are more open to accepting coupons compared to those heading home or to work.

**Actionable recommendations:**
- Target frequent coffee house visitors with priority coupon delivery
- Deliver coffee coupons in the morning hours for higher conversion
- Offer longer expiration windows (1 day) to increase acceptance
- Focus coupon delivery when drivers have no urgent destination and are with friends
- Prioritize younger demographics (under 30) in marketing efforts
